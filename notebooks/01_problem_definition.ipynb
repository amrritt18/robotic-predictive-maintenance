# Robotic Predictive Maintenance Using Machine Learning

## Phase 1: Problem Definition

### Dataset
AI4I 2020 Predictive Maintenance Dataset

Source: https://archive.ics.uci.edu/dataset/601/ai4i+2020+predictive+maintenance+dataset

### Objective
Develop a machine learning classification system that predicts whether a machine failure is present at the current observation using available operating measurements.

### Target Variable
`Machine failure`
- 0: No machine failure
- 1: Machine failure

### Project Type
Supervised Binary Classification


## 1. Business Problem

Unexpected machine failures can interrupt industrial production, increase maintenance costs, and reduce equipment availability.

The objective of this project is to develop a machine learning model that identifies observations associated with machine failure based on machine operating conditions.

The model will support maintenance decision-making by helping identify potentially problematic operating conditions.

The project uses the AI4I 2020 Predictive Maintenance Dataset, which contains synthetic industrial machine data.


## 2. Prediction Objective and Assumptions

### Prediction Objective
Predict the binary `Machine failure` label for the current observation.

### Assumptions
1. The five operating measurements and product type are available when a prediction is requested.
2. The target is the overall machine failure label, not an individual failure mode.
3. UDI and Product ID are excluded from the initial model.
4. TWF, HDF, PWF, OSF, and RNF are excluded from model inputs because they directly encode failure-mode labels.
5. The dataset represents synthetic observations and may not fully represent real industrial operating conditions.
6. Predictions indicate failure classification at the observed operating condition; they do not estimate remaining useful life or time until failure.


## 3. Initial Feature Selection

### Input Features
- Type
- Air temperature [K]
- Process temperature [K]
- Rotational speed [rpm]
- Torque [Nm]
- Tool wear [min]

### Target
- Machine failure

### Excluded Features
- UDI: Unique observation identifier
- Product ID: Product serial identifier
- TWF: Tool wear failure label
- HDF: Heat dissipation failure label
- PWF: Power failure label
- OSF: Overstrain failure label
- RNF: Random failure label

Feature selection may be revisited after EDA, but changes must be justified and documented.


## 4. Evaluation Strategy

The project will prioritize balancing missed failures and false alarms.

### Primary Evaluation Metrics
- Recall: Measures the proportion of actual failures detected.
- Precision: Measures the proportion of predicted failures that are actual failures.
- F1-score: Balances precision and recall.
- PR-AUC: Evaluates precision-recall performance across thresholds.

### Additional Metrics
- Confusion matrix
- Accuracy
- ROC-AUC

### Model Selection
The final model will not be selected using accuracy alone.

Recall, precision, false-alarm frequency, and overall operational usefulness will be considered together.

The classification threshold may be optimized using validation data. The test set will be reserved for final evaluation.


## 5. Project Limitations

1. The dataset is synthetic rather than collected directly from a deployed industrial robot.
2. The observations do not establish a reliable time-to-failure or remaining-useful-life target.
3. The dataset may not capture all real-world operating conditions, sensor noise, or changes over time.
4. The cost of missed failures and false alarms is not explicitly provided, so operational trade-offs must be assessed using available metrics.
5. Model performance on this dataset does not guarantee performance on real industrial equipment.
6. External validation on representative real-world data would be necessary before operational deployment.
